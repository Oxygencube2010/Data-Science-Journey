```python
import random
import matplotlib.pyplot as plt

def simulate_coin_flips(num_flips):
    """
    Simulates a series of coin flips and calculates the running proportion of heads
    """
    results = []
    heads_count = 0
    for i in range(1, num_flips +1):
        if random.random() < 0.5 : # simulate a fair coin
            heads_count += 1
        prob_heads = heads_count / i
        results.append(prob_heads)
    return results

def visualize_law_of_large_numbers(max_flips):
    """
    Visualizes the Law of Large Numbers by plotting the proportion of heads over an increasing number of coin flips
    """
    flip_counts = list(range(1,max_flips+1))
    prob = simulate_coin_flips(max_flips)

    plt.figure(figsize=(10,6))
    plt.plot(flip_counts, prob , label='Probability of Heads', color='blue')
    plt.axhline(0.5, color='red', linestyle='--', label='Expected Probability (0.5)')
    
    plt.title('Law of Large Numbers: Coin Flip Simulation')
    plt.xlabel('Number of Flips')
    plt.ylabel('Probability of Heads')
    plt.grid(True)
    plt.legend()
    plt.show()

# Set the maximum number of coin flips for the simulation
max_flips_to_simulate = 1000

# run simulations
visualize_law_of_large_numbers(max_flips_to_simulate)
```
